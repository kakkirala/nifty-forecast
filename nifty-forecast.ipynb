```python
from unsloth import FastLanguageModel
import torch
max_seq_length = 2048
dtype = None
load_in_4bit = True

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Qwen2.5-0.5B-Instruct",
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)

model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",
    use_gradient_checkpointing = "unsloth",
    random_state = 3407,
    use_rslora = False,
    loftq_config = None,
)
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("debashis74017/nifty-50-minute-data")

print("Path to dataset files:", path)
```

```python
import os
for f in sorted(os.listdir(path)):
    size_mb = os.path.getsize(os.path.join(path, f)) / 1e6
    print(f"{f}  ({size_mb:.1f} MB)")
```

```python
import pandas as pd

df = pd.read_csv(path + "/NIFTY 50_day.csv")
print("Rows, columns:", df.shape)
print("Columns:", df.columns.tolist())
print(df.head())
print(df.tail())
```

```python
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

# Clues from TODAY
df["ma5"] = df["close"].rolling(5).mean()        # 5-day average
df["ma10"] = df["close"].rolling(10).mean()      # 10-day average
df["pct_change"] = df["close"].pct_change()      # today's % move
df["day_range"] = (df["high"] - df["low"]) / df["close"]

# The ANSWER: tomorrow's close, pulled up to today's row
df["target"] = df["close"].shift(-1)

df = df.dropna().reset_index(drop=True)
print("Rows after features:", df.shape)
print(df[["date", "close", "ma5", "ma10", "target"]].head(12))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

features = ["close", "ma5", "ma10", "pct_change", "day_range"]

split = int(len(df) * 0.8)
train, test = df.iloc[:split], df.iloc[split:]
print("Train:", train["date"].min().date(), "→", train["date"].max().date())
print("Test: ", test["date"].min().date(), "→", test["date"].max().date())

model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
model.fit(train[features], train["target"])

pred = model.predict(test[features])
mae = mean_absolute_error(test["target"], pred)
rmse = np.sqrt(mean_squared_error(test["target"], pred))
print(f"\nMAE:  {mae:.2f} points  (average miss)")
print(f"RMSE: {rmse:.2f} points")
print(f"MAE as % of price: {mae/test['target'].mean()*100:.2f}%")
```

```python
# --- Predict tomorrow's % move, not the price ---
d2 = df[["date", "close"]].copy()
d2["ret1"] = d2["close"].pct_change(1)                  # today's % move
d2["ret5"] = d2["close"].pct_change(5)                  # last 5 days' % move
d2["vol5"] = d2["ret1"].rolling(5).std()                # recent jumpiness
d2["ma5_ratio"] = d2["close"] / d2["close"].rolling(5).mean() - 1   # distance from 5-day avg
d2["target"] = d2["ret1"].shift(-1)                     # TOMORROW's % move
d2 = d2.dropna().reset_index(drop=True)

features = ["ret1", "ret5", "vol5", "ma5_ratio"]
split = int(len(d2) * 0.8)
train, test = d2.iloc[:split], d2.iloc[split:]

model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
model.fit(train[features], train["target"])
pred = model.predict(test[features])

mae = mean_absolute_error(test["target"], pred)
print(f"MAE: {mae*100:.3f}%  (avg miss on daily move)")
print(f"Typical daily move: {test['target'].abs().mean()*100:.3f}%")
```

```python

import numpy as np
naive_mae = np.abs(test["target"]).mean()
print(f"Naive (always predict 0%): {naive_mae*100:.3f}%")
print(f"Our model:                 {mae*100:.3f}%")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

d2["up"] = (d2["target"] > 0).astype(int)   # 1 = tomorrow up, 0 = down
train, test = d2.iloc[:split], d2.iloc[split:]

clf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
clf.fit(train[features], train["up"])
pred_up = clf.predict(test[features])

acc = accuracy_score(test["up"], pred_up)
print(f"Direction accuracy: {acc*100:.1f}%")
print(f"Coin flip: ~50%")
print(f"Always predicting UP: {test['up'].mean()*100:.1f}%")
```
